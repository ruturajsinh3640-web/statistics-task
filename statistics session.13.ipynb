```python
import scipy.stats as stats
import numpy as np

# Sample delivery times in minutes
delivery_times = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]
hypothesized_mean = 30

# Perform one-sample t-test
t_statistic, p_value = stats.ttest_1samp(delivery_times, hypothesized_mean)

# Calculate descriptive statistics
mean_val = np.mean(delivery_times)
std_val = np.std(delivery_times, ddof=1)

print(f"Sample Mean: {mean_val}")
print(f"Sample Std Dev: {std_val:.3f}")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
import numpy as np
import scipy.stats as stats

# Set seed for reproducibility
np.random.seed(42)

# Simulate daily step counts for 30 users in each group
steps_before = np.random.normal(loc=6000, scale=1200, size=30)
steps_after = np.random.normal(loc=7500, scale=1500, size=30)

# Clean data (ensure valid step counts)
steps_before = np.clip(steps_before, 1000, None).astype(int)
steps_after = np.clip(steps_after, 1000, None).astype(int)

# Perform independent two-sample t-test (testing if 'after' > 'before')
t_stat, p_val = stats.ttest_ind(steps_after, steps_before, alternative='greater')

# Calculate descriptive statistics
print(f"Before - Mean: {np.mean(steps_before):.2f}, Std: {np.std(steps_before, ddof=1):.2f}")
print(f"After  - Mean: {np.mean(steps_after):.2f}, Std: {np.std(steps_after, ddof=1):.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.6f}")
```

```python
import numpy as np
import scipy.stats as stats

# Set random seed for reproducibility
np.random.seed(42)

# Simulate paired data for 10 users (screen time in hours)
screen_time_before = np.random.normal(loc=5.5, scale=1.2, size=10)
# Simulate 'after' values as generally lower than 'before'
screen_time_after = screen_time_before - np.random.normal(loc=1.3, scale=0.4, size=10)

# Ensure realistic constraints
screen_time_before = np.clip(screen_time_before, 1.0, 12.0)
screen_time_after = np.clip(screen_time_after, 0.5, 10.0)

# Perform paired t-test (testing if 'before' > 'after')
t_stat, p_val = stats.ttest_rel(screen_time_before, screen_time_after, alternative='greater')

# Descriptive statistics
print(f"Before - Mean: {np.mean(screen_time_before):.2f}h, Std: {np.std(screen_time_before, ddof=1):.2f}h")
print(f"After  - Mean: {np.mean(screen_time_after):.2f}h, Std: {np.std(screen_time_after, ddof=1):.2f}h")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.6f}")
```

```python
import numpy as np
import scipy.stats as stats

# Set random seed for reproducibility
np.random.seed(42)

# Simulate 15 random user ratings (on a 1 to 5 scale) for Zomato and Swiggy
zomato_ratings = np.random.normal(loc=4.2, scale=0.6, size=15)
swiggy_ratings = np.random.normal(loc=3.9, scale=0.7, size=15)

# Clip ratings to stay strictly within a realistic 1.0 to 5.0 rating scale
zomato_ratings = np.clip(zomato_ratings, 1.0, 5.0)
swiggy_ratings = np.clip(swiggy_ratings, 1.0, 5.0)

# Perform Welch's independent two-sample t-test
t_stat, p_val = stats.ttest_ind(zomato_ratings, swiggy_ratings, equal_var=False)

# Display descriptive stats and test results
print(f"Zomato Mean Rating: {np.mean(zomato_ratings):.2f} (Std: {np.std(zomato_ratings, ddof=1):.2f})")
print(f"Swiggy Mean Rating: {np.mean(swiggy_ratings):.2f} (Std: {np.std(swiggy_ratings, ddof=1):.2f})")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.4f}")
```

```python

import scipy.stats as stats
import numpy as np

# Sample data: Daily UPI transactions for 12 users on PhonePe
upi_transactions = [4, 6, 5, 7, 3, 5, 6, 8, 4, 5, 6, 7]
hypothesized_mean = 5.0

# Perform the one-sample t-test
t_statistic, p_value = stats.ttest_1samp(upi_transactions, hypothesized_mean)

# Calculate descriptive statistics
mean_val = np.mean(upi_transactions)
std_val = np.std(upi_transactions, ddof=1)

print(f"Sample Mean: {mean_val:.2f}")
print(f"Sample Standard Deviation: {std_val:.2f}")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```
