```python
import statistics

order_amounts = [250, 400, 350, 300, 500]

# Population variance and standard deviation
variance = statistics.pvariance(order_amounts)
standard_deviation = statistics.pstdev(order_amounts)

print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```

```python
import statistics

data_usage = [3.2, 2.8, 3.5, 3.0, 2.9, 3.1]

mean = statistics.mean(data_usage)
standard_deviation = statistics.pstdev(data_usage)

cv = (standard_deviation / mean) * 100

print("Mean:", round(mean, 2), "GB")
print("Standard Deviation:", round(standard_deviation, 2), "GB")
print("Coefficient of Variation:", round(cv, 2), "%")
```

```python
import statistics

likes = [120, 150, 130, 110, 170, 140, 135, 160, 125, 155]

mean = statistics.mean(likes)
standard_deviation = statistics.pstdev(likes)

print("Mean:", mean)
print("Standard Deviation:", round(standard_deviation, 2))
```
