```python
import random

def select_survey_participants(total_users=200, sample_size=20):
    """
    Simulates selecting a random sample of users for a feedback survey.
    """
    # Step 1: Generate a simulated list of user IDs (e.g., user_001 to user_200)
    user_pool = [f"user_{i:03d}" for i in range(1, total_users + 1)]
    
    # Step 2: Randomly sample 20 unique users from the pool
    # random.sample ensures no duplicates are chosen
    selected_users = random.sample(user_pool, sample_size)
    
    return user_pool, selected_users

# Run the simulation
if __name__ == "__main__":
    pool, sample = select_survey_participants(200, 20)
    
    print(f"🎵 Total User Pool Size: {len(pool)}")
    print(f"📊 Survey Sample Size: {len(sample)}\n")
    print("Selected User IDs for Feedback Survey:")
    for index, user_id in enumerate(sample, 1):
        print(f"  {index:2d}. {user_id}")
```

```python
import random
from collections import defaultdict, Counter

def generate_stratified_sample(orders, target_sample_size=60):
    """
    Generates a stratified sample of orders ensuring proportional representation by city.
    """
    total_orders = len(orders)
    
    # Step 1: Group orders by city
    city_groups = defaultdict(list)
    for order in orders:
        city_groups[order['city']].append(order)
        
    stratified_sample = []
    allocation_summary = {}

    # Step 2: Calculate proportions and sample accordingly
    for city, city_orders in city_groups.items():
        city_count = len(city_orders)
        proportion = city_count / total_orders
        
        # Calculate target sample size for this city
        # Using round() to get the nearest integer
        city_sample_size = round(proportion * target_sample_size)
        
        # Ensure we don't try to sample more than available, and handle edge cases
        city_sample_size = max(1, min(city_sample_size, city_count))
        
        # Randomly sample from this specific city group
        sampled_subset = random.sample(city_orders, city_sample_size)
        stratified_sample.extend(sampled_subset)
        
        allocation_summary[city] = {
            "total_in_pool": city_count,
            "actual_percentage": proportion * 100,
            "sampled_count": city_sample_size
        }
        
    return stratified_sample, allocation_summary

# --- Simulation Setup ---
if __name__ == "__main__":
    # Mock cities in Gujarat
    cities = ['Ahmedabad', 'Surat', 'Vadodara', 'Rajkot']
    # Define weight distribution for the 1000 orders (e.g., Ahmedabad & Surat have more volume)
    city_weights = [0.45, 0.30, 0.15, 0.10] 
    
    # Generate 1000 simulated Flipkart orders
    order_pool = []
    for i in range(1, 1001):
        city = random.choices(cities, weights=city_weights, k=1)[0]
        order_pool.append({"order_id": f"ORD_{i:04d}", "city": city})
        
    # Run the stratified sampling function
    sample, summary = generate_stratified_sample(order_pool, target_sample_size=60)
    
    # --- Display Results ---
    print(f"📦 Total Order Pool: {len(order_pool)} orders")
    print(f"🎯 Target Sample Size: 60 orders | Actual Sampled: {len(sample)} orders\n")
    print("City Breakdown & Stratified Allocation:")
    print("-" * 55)
    print(f"{'City':<12} | {'Pool Count':<10} | {'Pool %':<8} | {'Sampled'}")
    print("-" * 55)
    
    for city, data in summary.items():
        print(f"{city:<12} | {data['total_in_pool']:<10} | {data['actual_percentage']:>5.1f}%   | {data['sampled_count']}")
    print("-" * 55)
```

```python
import random

def systematic_sampling(reviews, interval=10):
    """
    Performs systematic sampling by picking a random start index 
    and selecting every 10th review.
    """
    # Step 1: Pick a random starting index between 0 and interval - 1
    start_index = random.randint(0, interval - 1)
    
    # Step 2: Slice the list taking every 'interval'-th review
    sampled_reviews = reviews[start_index::interval]
    
    return start_index, sampled_reviews

# --- Simulation Setup ---
if __name__ == "__main__":
    # Generate 500 mock Zomato reviews
    zomato_reviews = [f"Review #{i}: Great food!" for i in range(1, 501)]
    
    # Run the systematic sampling function (interval = 10)
    start, sample = systematic_sampling(zomato_reviews, interval=10)
    
    # --- Display Results ---
    print(f"🍽️ Total Reviews in Pool: {len(zomato_reviews)}")
    print(f"🎲 Random Start Index: {start}")
    print(f"📋 Total Sampled Reviews: {len(sample)} (Every 10th review)\n")
    
    print("First 5 sampled reviews:")
    for rev in sample[:5]:
        print(f"  - {rev}")
```

```python
import numpy as np

# Set seed for reproducible results
np.random.seed(42)

# Population Parameters
N = 10000            # Total delivery records
true_std = 10.0      # Assumed standard deviation in minutes
sample_sizes = [50, 200, 500]
z_score = 1.96       # 95% confidence level

# Step 1: Simulate 10,000 Swiggy delivery times (centered around 35 mins)
population = np.random.normal(loc=35.0, scale=true_std, size=N)
population = np.maximum(population, 5.0) # Ensure no negative times

print(f"📦 Population Size: {N} delivery records")
print(f"📈 Population Average (Simulated): {np.mean(population):.2f} mins\n")

print(f"{'Sample Size (n)':<15} | {'Sample Mean':<12} | {'Std Error (SE)':<15} | {'95% Margin of Error'}")
print("-" * 70)

# Step 2: Calculate metrics for each sample size
for n in sample_sizes:
    # Take a random sample without replacement
    sample = np.random.choice(population, size=n, replace=False)
    
    sample_mean = np.mean(sample)
    sample_std = np.std(sample, ddof=1) # Sample standard deviation
    
    # Finite Population Correction (FPC) factor
    fpc = np.sqrt((N - n) / (N - 1))
    
    # Standard Error with FPC
    se = (sample_std / np.sqrt(n)) * fpc
    
    # Margin of Error
    moe = z_score * se
    
    print(f"{n:<15} | {sample_mean:<12.2f} | {se:<15.2f} | ±{moe:.2f} mins")

print("-" * 70)
```

```python
import random

def generate_ipl_stratified_sample(fan_pool, target_sample_size=40):
    """
    Performs stratified sampling on IPL fans using standard dictionaries 
    and proportional allocation.
    """
    total_fans = len(fan_pool)
    
    # Step 1: Custom grouping logic without defaultdict or pandas
    fans_by_team = {}
    for fan in fan_pool:
        team = fan['favorite_team']
        if team not in fans_by_team:
            fans_by_team[team] = []
        fans_by_team[team].append(fan)
        
    stratified_sample = []
    allocation_report = {}

    # Step 2: Calculate proportions and sample proportionally from each team group
    for team, members in fans_by_team.items():
        team_count = len(members)
        proportion = team_count / total_fans
        
        # Calculate quota for this team (rounded to nearest integer)
        quota = round(proportion * target_sample_size)
        
        # Ensure quota doesn't exceed available members or drop below 1 (if team exists)
        quota = max(1, min(quota, team_count))
        
        # Randomly sample the quota from this specific team's members
        sampled_subset = random.sample(members, quota)
        stratified_sample.extend(sampled_subset)
        
        allocation_report[team] = {
            "total_pool": team_count,
            "percentage": proportion * 100,
            "sample_quota": quota
        }
        
    return stratified_sample, allocation_report

# --- Simulation Setup ---
if __name__ == "__main__":
    # Popular IPL teams
    ipl_teams = ['Chennai Super Kings', 'Mumbai Indians', 'Royal Challengers Bengaluru', 'Gujarat Titans', 'Kolkata Knight Riders']
    team_weights = [0.30, 0.25, 0.20, 0.15, 0.10]
    
    # Generate a pool of 500 IPL fans
    all_fans = []
    for i in range(1, 501):
        fav_team = random.choices(ipl_teams, weights=team_weights, k=1)[0]
        all_fans.append({"fan_id": f"FAN_{i:03d}", "favorite_team": fav_team})
        
    # Run the custom sampling function
    sample_set, report = generate_ipl_stratified_sample(all_fans, target_sample_size=40)
    
    # --- Display Results ---
    print(f"🏟️ Total Fan Pool: {len(all_fans)} fans")
    print(f"🎯 Target Survey Sample: 40 fans | Actual Sampled: {len(sample_set)}\n")
    print(f"{'Team Name':<30} | {'Pool Count':<10} | {'Share (%)':<10} | {'Sampled'}")
    print("-" * 65)
    
    for team, data in report.items():
        print(f"{team:<30} | {data['total_pool']:<10} | {data['percentage']:>5.1f}%    | {data['sample_quota']}")
    print("-" * 65)
```
