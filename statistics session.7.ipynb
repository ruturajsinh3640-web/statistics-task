```python
def bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email):
    """
    Computes the probability of churn given an email was received using Bayes' Theorem:
    P(Churn | Email) = [ P(Email | Churn) * P(Churn) ] / P(Email)
    """
    posterior_probability = (prob_email_given_churn * prior_churn) / prob_email
    return posterior_probability

# Test the function with the given values
prior_churn = 0.15
prob_email_given_churn = 0.6
prob_email = 0.25

result = bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email)

print(f"Probability of churn given the email: {result:.4f} ({result * 100:.2f}%)")
```

```python
# Define given probabilities
p_f = 0.02                  # Probability of fraud
p_not_f = 1 - p_f           # Probability of normal transaction
p_a_given_f = 0.90          # True positive rate (correctly flagged fraud)
p_a_given_not_f = 0.05      # False positive rate (falsely flagged normal)

# Calculate total probability of an alert P(A)
p_a = (p_a_given_f * p_f) + (p_a_given_not_f * p_not_f)

# Apply Bayes' Theorem to find P(F | A)
p_f_given_a = (p_a_given_f * p_f) / p_a

print(f"Probability of actual fraud given an alert: {p_f_given_a:.4f} ({p_f_given_a * 100:.2f}%)")
```

```python
def bayes_recommendation_probability(prior_craving, prob_rain_given_craving, prob_rain):
    """
    Computes P(Craving | Rain) using Bayes' Theorem.
    """
    posterior = (prob_rain_given_craving * prior_craving) / prob_rain
    return posterior

# Test with hypothetical delivery app metrics
prior_craving = 0.25         # General probability a user wants comfort food
prob_rain_given_craving = 0.80 # Probability it's raining given they want comfort food
prob_rain = 0.35             # Overall probability of rain in the city today

result = bayes_recommendation_probability(prior_craving, prob_rain_given_craving, prob_rain)

print(f"Probability of craving comfort food given rain: {result:.4f} ({result * 100:.2f}%)")
```
