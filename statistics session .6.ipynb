```python
total_songs = 8
arijit_songs = 3

# Calculate songs not by Arijit Singh
not_arijit = total_songs - arijit_songs

# Calculate probability
probability = not_arijit / total_songs

print(f"Probability of not playing an Arijit Singh song: {probability} ({probability * 100}%)")
```

```python
# Define the deck of cards
total_cards = list(range(1, 11))

# Find even-numbered cards
even_cards = [card for card in total_cards if card % 2 == 0]

# Calculate probability
probability = len(even_cards) / len(total_cards)

print(f"Total cards: {len(total_cards)}")
print(f"Even cards: {even_cards}")
print(f"Probability of drawing an even card: {probability} ({probability * 100}%)")
```

```python
# Define probabilities
p_pizza = 0.3
p_burger = 0.4
p_both = 0.1

# Apply the addition rule
p_pizza_or_burger = p_pizza + p_burger - p_both

print(f"Probability of ordering Pizza or Burger: {p_pizza_or_burger} ({p_pizza_or_burger * 100}%)")
```
