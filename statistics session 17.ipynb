```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

ad_spend = np.array([5.2, 10.5, 15.0, 22.3, 28.1, 35.0, 40.2, 45.6, 50.0, 58.4])
downloads = 15000 + 3200 * ad_spend + np.random.normal(0, 8000, size=len(ad_spend))


downloads = np.clip(downloads, 5000, 200000).astype(int)

df = pd.DataFrame({
    'Ad_Spend_Lakhs': ad_spend,
    'App_Downloads': downloads
})


m, c = np.polyfit(df['Ad_Spend_Lakhs'], df['App_Downloads'], 1)

plt.figure(figsize=(8, 5))
plt.scatter(df['Ad_Spend_Lakhs'], df['App_Downloads'], color='crimson', label='Sample Data Points', s=70, zorder=5)

plt.plot(df['Ad_Spend_Lakhs'], m * df['Ad_Spend_Lakhs'] + c, color='dodgerblue', linewidth=2, label=f'Line of Best Fit\n(y = {m:.2f}x + {c:.2f})', zorder=3)

plt.title('Myntra Ad Spend vs. App Downloads', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Ad Spend (in Lakhs INR)', fontsize=12)
plt.ylabel('Number of App Downloads', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# Save plot
plt.savefig('myntra_scatter_plot.png', dpi=300)
plt.close()

print(df)
print(f"Slope (m): {m}, Intercept (c): {c}")
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Sample data: Daily advertising spend (in lakhs) and daily orders
ad_spend = np.array([2.0, 4.5, 6.0, 8.5, 10.0, 12.5, 15.0, 18.0, 20.0, 25.0])
orders = np.array([2450, 3600, 4200, 5800, 6500, 7900, 9100, 10800, 11700, 14300])

# Reshape X for scikit-learn (requires 2D array)
X = ad_spend.reshape(-1, 1)
y = orders

# Initialize and fit the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Output coefficients
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (Coefficient): {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Sample data from previous turn
ad_spend = np.array([2.0, 4.5, 6.0, 8.5, 10.0, 12.5, 15.0, 18.0, 20.0, 25.0])
orders = np.array([2450, 3600, 4200, 5800, 6500, 7900, 9100, 10800, 11700, 14300])

X = ad_spend.reshape(-1, 1)
y = orders

model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

r2 = r2_score(y, y_pred)
print(f"R-squared: {r2}")
```
