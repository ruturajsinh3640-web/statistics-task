```python
import numpy as np
from scipy import stats

# 1. Sample data: Average daily orders before and after the cashback offer
np.random.seed(42)  # For reproducibility
orders_before = np.random.normal(loc=1200, scale=150, size=30)  # 30 days before
orders_after = np.random.normal(loc=1350, scale=160, size=30)   # 30 days after

# 2. Perform independent two-sample t-test
# alternative='greater' directly tests if orders_after > orders_before in SciPy >= 1.6.0
t_stat, p_value_two_tailed = stats.ttest_ind(orders_after, orders_before, equal_var=False)

# For a one-tailed test (H1: after > before):
# If the t-statistic is positive, divide the two-tailed p-value by 2.
if t_stat > 0:
    p_value_one_tailed = p_value_two_tailed / 2
else:
    p_value_one_tailed = 1 - (p_value_two_tailed / 2)

# 3. Print Results
alpha = 0.05
print(f"T-statistic: {t_stat:.4f}")
print(f"One-tailed p-value: {p_value_one_tailed:.4f}")

if p_value_one_tailed < alpha:
    print("Result: Reject the null hypothesis ($H_0$). There is significant evidence that the cashback offer increased daily orders.")
else:
    print("Result: Fail to reject the null hypothesis ($H_0$). There is not enough evidence to claim an increase in daily orders.")
```

```python

import numpy as np
from scipy import stats

# 1. Sample data: Listening times (in minutes) of users after the new algorithm
np.random.seed(42)  # For reproducibility
listening_times = np.random.normal(loc=43, scale=8, size=50) 

# Baseline mean from before the algorithm
mu_hypothesized = 40

# 2. Perform one-sample one-tailed t-test
# alternative='greater' tests if the true mean > mu_hypothesized
t_stat, p_value = stats.ttest_1samp(listening_times, mu_hypothesized, alternative='greater')

# 3. Print Results
alpha = 0.05
print(f"Sample Mean Listening Time: {np.mean(listening_times):.2f} minutes")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < alpha:
    print("Result: Reject the null hypothesis ($H_0$). Evidence suggests the algorithm increases listening time.")
else:
    print("Result: Fail to reject the null hypothesis ($H_0$). Not enough evidence to claim an increase.")
```

```python
import numpy as np
from scipy import stats

# 1. Simulate A/B test data (e.g., daily purchases per user)
np.random.seed(42)  # For reproducibility
control_group = np.random.normal(loc=1.20, scale=0.4, size=1000)   # Old page (no banner)
treatment_group = np.random.normal(loc=1.26, scale=0.4, size=1000) # New page (with discount banner)

# 2. Perform independent two-sample t-test
t_stat, p_value_two_tailed = stats.ttest_ind(treatment_group, control_group)

# Adjust for a one-tailed test (H1: treatment > control)
if t_stat > 0:
    p_value = p_value_two_tailed / 2
else:
    p_value = 1 - (p_value_two_tailed / 2)

# 3. Decision making at 95% confidence level
alpha = 0.05
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < alpha:
    print("Result: Reject the null hypothesis ($H_0$). The discount banner effectively increases purchases!")
else:
    print("Result: Fail to reject the null hypothesis ($H_0$). Not enough evidence to roll out the banner.")
```
