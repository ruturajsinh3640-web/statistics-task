```python
import os
import pandas as pd

# 1. Check if the file exists, if not, create a sample dataset for testing
file_name = 'sales_data.csv'

if not os.path.exists(file_name):
    print(f"'{file_name}' not found. Creating a sample dataset for you...")
    sample_data = {
        'Country': ['USA', 'Canada', 'USA', 'Germany', 'France', 'Canada', 'Germany', 'USA', 'France', 'Germany'],
        'Quantity': [10, 5, 8, 12, 7, 3, 15, 20, 6, 9],
        'UnitPrice': [15.5, 20.0, 10.0, 25.0, 30.0, 45.0, 12.0, 8.5, 50.0, 22.0]
    }
    df_sample = pd.DataFrame(sample_data)
    df_sample.to_csv(file_name, index=False)

# 2. Load the dataset
df = pd.read_csv(file_name)

# 3. Calculate Total Sales for each row
df['TotalSales'] = df['Quantity'] * df['UnitPrice']

# 4. Create the Pivot Table equivalent (Total Sales by Country)
pivot_df = pd.pivot_table(
    df,
    values='TotalSales',
    index=['Country'],
    aggfunc='sum'
).reset_index()

# Sort the table in descending order to easily find the top countries
pivot_df = pivot_df.sort_values(by='TotalSales', ascending=False)

# 5. Identify the top 3 countries by total sales
top_3_countries = pivot_df.head(3)['Country'].tolist()
top_3_text = "Top 3 Countries: " + ", ".join(top_3_countries)

# 6. Write the table and the top 3 summary to an Excel file
output_file = 'sales_analysis.xlsx'

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    pivot_df.to_excel(writer, sheet_name='Sales Summary', index=False, startrow=0)
    
    ws = writer.sheets['Sales Summary']
    row_below_table = len(pivot_df) + 3
    
    ws.cell(row=row_below_table, column=1, value=top_3_text)

print(f"\nPivot analysis successfully saved to '{output_file}'!")
print(top_3_text)
```

```python

# Example DataFrame (tamemari file mate pd.read_csv('your_file.csv') use karjo)
data = {
    'InvoiceNo': ['INV001', 'INV001', 'INV002', 'INV003', 'INV003'],
    'CustomerName': ['Rahul', 'Rahul', 'Priya', 'Amit', 'Amit'],
    'Sales': [4000, 7000, 12000, 3000, 4000]
}
df = pd.DataFrame(data)

# 1. InvoiceNo mujab Total Sales (Order Value) calculate karo
order_value_df = df.groupby(['InvoiceNo', 'CustomerName'])['Sales'].sum().reset_index()
order_value_df.rename(columns={'Sales': 'OrderValue'}, inplace=True)

# 2. Badha orders ni Average Order Value calculate karo
avg_order_value = order_value_df['OrderValue'].mean()
print(f"Average Order Value: ₹{avg_order_value:.2f}\n")

# 3. ₹10,000 thi vadhare na Orders ne highlight (filter) karo
high_value_orders = order_value_df[order_value_df['OrderValue'] > 10000]

print("Orders Above ₹10,000:")
print(high_value_orders)
```

```python

from scipy import stats

# Example DataFrame (Tamari file mate pd.read_csv('your_file.csv') use karjo)
data = {
    'Country': ['France', 'Germany', 'France', 'Germany', 'France', 'Germany', 'France', 'Germany'],
    'OrderValue': [12000, 8500, 15000, 9200, 11000, 7800, 13500, 8900]
}
df = pd.DataFrame(data)

# 1. France ane Germany na Order Values ne alag karo
france_orders = df[df['Country'] == 'France']['OrderValue']
germany_orders = df[df['Country'] == 'Germany']['OrderValue']

# 2. Independent Two-Sample T-Test run karo (Unequal Variances mate equal_var=False)
t_stat, p_value = stats.ttest_ind(france_orders, germany_orders, equal_var=False)

print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}\n")

# 3. Significance check karo (alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print("Result: The difference is statistically significant. (France ane Germany na average order value ma mota/khara bhav fark che).")
else:
    print("Result: The difference is NOT statistically significant. (Je fark che te random chance kari shake che).")
```

```python


# Example DataFrame (Tamari actual file mate pd.read_csv('your_file.csv') use karjo)
data = {
    'InvoiceNo': ['INV001', 'INV001', 'INV002', 'INV003', 'INV004', 'INV005'],
    'Country': ['France', 'France', 'Germany', 'France', 'Germany', 'Germany'],
    'Sales': [6000, 6000, 8500, 15000, 9200, 11000]
}
df = pd.DataFrame(data)

# 1. Invoice mujab total Order Value calculate karo
order_values = df.groupby(['InvoiceNo', 'Country'])['Sales'].sum().reset_index()
order_values.rename(columns={'Sales': 'OrderValue'}, inplace=True)

# 2. Country-wise Insights (Average order size, total sales, order count) calculate karo
insights_df = order_values.groupby('Country')['OrderValue'].agg(
    Average_Order_Size='mean',
    Total_Sales='sum',
    Total_Orders='count'
).reset_index()

# 3. ₹10,000 thi vadhare na high-value orders filter karo
high_value_count = order_values[order_values['OrderValue'] > 10000].shape[0]

print("--- Generated Insights ---")
print(insights_df)
print(f"\nTotal Orders above ₹10,000: {high_value_count}")

# 4. Excel file ma multiple sheets ('Orders' ane 'Insights') sathe save karo
with pd.ExcelWriter('Customer_Purchasing_Insights.xlsx', engine='openpyxl') as writer:
    order_values.to_excel(writer, sheet_name='OrderValues', index=False)
    insights_df.to_excel(writer, sheet_name='Insights', index=False)

print("\nSuccess: Summary 'Insights' sheet ma Excel file ma save thai gai che!")
```
