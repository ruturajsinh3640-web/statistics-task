```python
import numpy as np
import matplotlib.pyplot as plt

# Set a random seed for reproducibility
np.random.seed(42)

# 1. Generate a heavily skewed population (e.g., simulating Instagram likes)
# An exponential distribution naturally models heavily right-skewed data.
population_size = 100000
skewed_population = np.random.exponential(scale=50, size=population_size)

# 2. Simulation parameters
num_samples = 1000
sample_size = 30

# 3. Take 1000 random samples, each of size 30, and calculate their means
# Using a 2D array makes this efficient: 1000 rows (samples) x 30 columns (observations)
samples = np.random.choice(skewed_population, size=(num_samples, sample_size))
sample_means = np.mean(samples, axis=1)

# 4. Plot the histogram of the sample means
plt.figure(figsize=(10, 6))
count, bins, ignored = plt.hist(sample_means, bins=30, color='skyblue', edgecolor='black', density=True)

# Add styling and labels
plt.title('Demonstrating the Central Limit Theorem (Sample Size n = 30)', fontsize=14, fontweight='bold')
plt.xlabel('Sample Mean (Average Instagram Likes)', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(axis='y', alpha=0.75)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set a random seed for reproducibility
np.random.seed(42)

# 1. Generate a heavily skewed population (e.g., Zomato order amounts)
population_size = 100000
skewed_population = np.random.exponential(scale=300, size=population_size)

# 2. Define different sample sizes to test
sample_sizes = [2, 10, 30, 100]
num_samples = 1000

# 3. Create a multi-panel plot (2x2 grid)
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, n in enumerate(sample_sizes):
    # Take 1000 samples of size 'n' and calculate their means
    samples = np.random.choice(skewed_population, size=(num_samples, n))
    sample_means = np.mean(samples, axis=1)
    
    # Plot histogram for this sample size
    axes[i].hist(sample_means, bins=30, color='skyblue', edgecolor='black', density=True)
    axes[i].set_title(f'Sample Size (n) = {n}', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Sample Mean', fontsize=10)
    axes[i].set_ylabel('Density', fontsize=10)
    axes[i].grid(axis='y', alpha=0.75)

# Overall layout adjustment
plt.suptitle('The Central Limit Theorem: Effect of Increasing Sample Size', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set a random seed for reproducibility
np.random.seed(42)

# 1. Generate a skewed population of Swiggy delivery times (in minutes)
# Most orders take ~25-30 mins, but a long tail represents delayed orders.
population_size = 100000
base_times = np.random.exponential(scale=18, size=population_size)
swiggy_population = base_times + 12 # Minimum delivery time of 12 minutes

# True population mean (for reference)
true_mean = np.mean(swiggy_population)

# 2. Simulation parameters based on the Swiggy scenario
num_samples = 1000
sample_size = 40 # Standard sample size for quality control checks

# 3. Take 1000 random samples and calculate their mean delivery times
samples = np.random.choice(swiggy_population, size=(num_samples, sample_size))
sample_means = np.mean(samples, axis=1)

# 4. Plot the distribution for business analytics
plt.figure(figsize=(10, 6))
plt.hist(sample_means, bins=30, color='orange', edgecolor='black', density=True, alpha=0.8)

# Add business decision lines
sla_threshold = 35.0  # Promised SLA limit
plt.axvline(sla_threshold, color='red', linestyle='--', linewidth=2, label=f'SLA Threshold ({sla_threshold} mins)')
plt.axvline(true_mean, color='green', linestyle='-', linewidth=2, label=f'True Population Mean ({true_mean:.1f} mins)')

# Styling and labels
plt.title('CLT in Action: Monitoring Swiggy Average Delivery Times (n = 40)', fontsize=14, fontweight='bold')
plt.xlabel('Average Sample Delivery Time (Minutes)', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.75)

# Show the plot
plt.tight_layout()
plt.show()

# Quick business insight calculation
percent_under_sla = np.mean(sample_means < sla_threshold) * 100
print(f"Percentage of sample batches meeting the {sla_threshold}-minute SLA: {percent_under_sla:.1f}%")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set a random seed for reproducibility
np.random.seed(42)

# 1. Generate a skewed population (e.g., Swiggy delivery times)
population_size = 100000
swiggy_population = np.random.exponential(scale=18, size=population_size) + 12

num_samples = 1000

# 2. Calculate sample means for n = 5 and n = 30
means_n5 = np.mean(np.random.choice(swiggy_population, size=(num_samples, 5)), axis=1)
means_n30 = np.mean(np.random.choice(swiggy_population, size=(num_samples, 30)), axis=1)

# 3. Plot side-by-side histograms for comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Left plot: n = 5
axes[0].hist(means_n5, bins=30, color='salmon', edgecolor='black', density=True)
axes[0].set_title('Small Sample Size (n = 5)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Sample Mean (Minutes)', fontsize=11)
axes[0].set_ylabel('Density', fontsize=11)
axes[0].grid(axis='y', alpha=0.75)

# Right plot: n = 30
axes[1].hist(means_n30, bins=30, color='skyblue', edgecolor='black', density=True)
axes[1].set_title('Standard Sample Size (n = 30)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Sample Mean (Minutes)', fontsize=11)
axes[1].set_ylabel('Density', fontsize=11)
axes[1].grid(axis='y', alpha=0.75)

# Overall layout
plt.suptitle('Central Limit Theorem: Impact of Sample Size', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# 1. Generate a heavily skewed non-normal population (Exponential distribution)
population = np.random.exponential(scale=2.0, size=100000)

# 2. Simulation parameters
num_samples = 1000
sample_size = 30

# 3. Take random samples and compute their means
sample_means = [np.mean(np.random.choice(population, size=sample_size)) for _ in range(num_samples)]

# 4. Plotting the results
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Original Population Plot
axes[0].hist(population, bins=50, color='salmon', edgecolor='black', density=True)
axes[0].set_title('Original Population (Exponentially Skewed)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Value', fontsize=11)
axes[0].set_ylabel('Density', fontsize=11)

# Sample Means Plot (CLT in effect)
axes[1].hist(sample_means, bins=30, color='skyblue', edgecolor='black', density=True)
axes[1].set_title('Sampling Distribution of Means (n = 30)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Sample Mean', fontsize=11)
axes[1].set_ylabel('Density', fontsize=11)

plt.tight_layout()
plt.show()
```
