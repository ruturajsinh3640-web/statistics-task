```python

# Daily step counts for 7 days
steps = [6500, 8200, 7100, 9500, 7800, 10200, 8800]

# Calculate the mean
mean_steps = sum(steps) / len(steps)

print("Daily step counts:", steps)
print("Mean (average) steps:", mean_steps)
```

```python

def find_median(times):
    times = sorted(times)
    n = len(times)

    if n % 2 == 1:
        return times[n // 2]
    else:
        return (times[n // 2 - 1] + times[n // 2]) / 2


delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

median = find_median(delivery_times)

print("Median delivery time:", median, "minutes")
```

```python
from collections import Counter

genres = [
    'Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog',
    'Tech', 'Music', 'Comedy', 'Vlog', 'Music', 'Comedy'
]

genre_count = Counter(genres)

mode = genre_count.most_common(1)[0][0]

print("Most watched genre:", mode)
```

```python
transactions = [100, 120, 105, 110, 5000, 115, 108]

mean = sum(transactions) / len(transactions)

sorted_transactions = sorted(transactions)
n = len(sorted_transactions)
median = sorted_transactions[n // 2]

print("Mean:", mean)
print("Median:", median)
```
