```python
import matplotlib.pyplot as plt
import numpy as np

# Set a random seed for reproducibility (optional, but good for consistent results)
np.random.seed(42)

# Parameters for the normal distribution
mean = 5000
std_dev = 1000
num_samples = 10000

# Generate 1000 random numbers following the normal distribution
data = np.random.normal(loc=mean, scale=std_dev, size=num_samples)

# Plot the distribution using a histogram
plt.figure(figsize=(10, 6))
plt.hist(
    data, bins=30, edgecolor="black", alpha=0.7, color="skyblue"
)  # 30 bins for smooth grouping

# Add labels and title
plt.title(
    "Normal Distribution (Mean = 500, Standard Deviation = 100)", fontsize=14
)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

# Show grid lines for easier reading
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Display the plot
plt.show()
```
