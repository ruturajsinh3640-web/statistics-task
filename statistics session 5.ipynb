```python
import numpy as np

# Sample list of 20 delivery times (in minutes)
delivery_times = [22, 35, 18, 42, 29, 31, 24, 50, 38, 27, 21, 33, 45, 26, 30, 39, 25, 36, 40, 28]

# Calculate the 25th, 50th, and 75th percentiles
p25, p50, p75 = np.percentile(delivery_times, [25, 50, 75])

print(f"25th Percentile: {p25} minutes")
print(f"50th Percentile (Median): {p50} minutes")
print(f"75th Percentile: {p75} minutes")
```

```python
import numpy as np

# Sample array of IPL player match scores
ipl_scores = np.array([12, 45, 82, 3, 105, 67, 34, 21, 56, 89, 14, 73, 40, 62, 19, 94])

# Calculate quartiles using np.quantile (0.25 for Q1, 0.50 for Q2, 0.75 for Q3)
q1, q2, q3 = np.quantile(ipl_scores, [0.25, 0.50, 0.75])

print(f"Q1 (25th Percentile): {q1}")
print(f"Q2 (50th Percentile / Median): {q2}")
print(f"Q3 (75th Percentile): {q3}")
```

```python
import numpy as np

# Sample daily step counts for 30 days
step_counts = [
    6500, 7200, 8100, 4300, 9500, 10200, 11000, 
    5400, 7800, 8200, 8600, 3100, 9200, 10500, 
    7600, 8300, 6900, 11500, 8000, 7500, 6400, 
    8900, 9400, 7100, 8200, 7700, 10100, 8400, 7900, 8800
]

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
q1, q3 = np.percentile(step_counts, [25, 75])

# Calculate IQR
iqr = q3 - q1

print(f"25th Percentile (Q1): {q1}")
print(f"75th Percentile (Q3): {q3}")
print(f"Interquartile Range (IQR): {iqr}")
```

```python
import numpy as np

def detect_order_outliers(order_amounts):
    # Convert list to numpy array for easy calculation
    data = np.array(order_amounts)
    
    # Calculate Q1 (25th percentile) and Q3 (75th percentile)
    q1, q3 = np.percentile(data, [25, 75])
    
    # Calculate Interquartile Range (IQR)
    iqr = q3 - q1
    
    # Define outlier boundaries
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    # Filter the outliers
    outliers = data[(data < lower_bound) | (data > upper_bound)]
    
    print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
    print(f"Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
    print(f"Detected Outlier Order Amounts: {list(outliers)}")
    
    return outliers

# Sample list of Swiggy order amounts (including some unusually low/high values)
swiggy_orders = [180, 250, 310, 290, 420, 220, 350, 1500, 300, 275, 410, 50, 330, 280, 2400]

detect_order_outliers(swiggy_orders)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample weekly screen time data for the past 8 weeks (in hours)
screen_time_hours = [28, 32, 30, 45, 31, 29, 33, 58]

# Create the boxplot
plt.figure(figsize=(8, 5))
box = plt.boxplot(screen_time_hours, 
                  patch_artist=True,
                  boxprops=dict(facecolor='lightblue', color='darkblue', linewidth=1.5),
                  medianprops=dict(color='red', linewidth=2),
                  flierprops=dict(marker='o', color='red', markerfacecolor='orange', markersize=10, label='Outliers'))

# Labels and styling
plt.title('Weekly Screen Time Distribution (Past 8 Weeks)', fontsize=14, fontweight='bold')
plt.ylabel('Screen Time (Hours)', fontsize=12)
plt.xticks([1], ['Past 8 Weeks'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```
