```python
import random

# Generate 20 random Zomato order values between 100 and 1000
order_amounts = [random.randint(100, 1000) for _ in range(20)]

# Print the data in a format ready to copy-paste into Excel or Google Sheets
print("Order Amount")
for amount in order_amounts:
    print(amount)
```

```python
import statistics

order_amounts = [150, 200, 150, 300, 450, 200, 150]

mean_val = statistics.mean(order_amounts)
median_val = statistics.median(order_amounts)
mode_val = statistics.mode(order_amounts)

print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Mode: {mode_val}")
```

```python
import statistics

population = [150, 200, 150, 300, 450, 200, 150]
sample = population[:5]

pop_mean = statistics.mean(population)
sample_mean = statistics.mean(sample)

print(f"Population: {population}")
print(f"Population Mean: {pop_mean}")
print(f"Sample (first 5): {sample}")
print(f"Sample Mean: {sample_mean}")
print(f"Difference: {sample_mean - pop_mean}")
```

```python
import pandas as pd
# Simulating a Zomato dataset
# Population: All orders ever placed
population_df = pd.DataFrame({'Order_Amount': [150, 200, 250, 300, 350, 400, 450, 500, 550, 600]})

# Sample: A smaller, practical subset (e.g., first 3 orders)
sample_df = population_df.sample(n=3, random_state=42)

print("Sample Mean:", sample_df['Order_Amount'].mean())
print("Population Mean:", population_df['Order_Amount'].mean())
```
