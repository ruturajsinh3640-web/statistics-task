```python
import pandas as pd

data = {
    'song_name': [
        'Blinding Lights', 'Watermelon Sugar', 'Levitating', 'Save Your Tears',
        'Peaches', 'Good 4 U', 'Stay', 'Industry Baby', 'Heat Waves',
        'Bad Guy', 'Old Town Road', 'Sunflower'
    ],
    'danceability': [0.514, 0.548, 0.702, 0.680, 0.677, 0.563, 0.591, 0.737, 0.759, 0.701, 0.878, 0.755],
    'energy': [0.730, 0.816, 0.825, 0.826, 0.696, 0.664, 0.764, 0.704, 0.522, 0.425, 0.619, 0.522],
    'popularity': [85, 82, 88, 80, 79, 86, 90, 84, 83, 78, 81, 87]
}

df = pd.DataFrame(data)
correlation = df['danceability'].corr(df['popularity'], method='pearson')

print("Dataset:")
print(df.to_string(index=False))
print(f"\nPearson Correlation Coefficient (Danceability vs Popularity): {correlation:.4f}")
```

```python
import pandas as pd
from scipy.stats import spearmanr

# Creating a realistic sample dataset of IPL players (runs and Instagram followers in millions)
data = {
    'player_name': [
        'Virat Kohli', 'MS Dhoni', 'Rohit Sharma', 'Hardik Pandya', 
        'Ravindra Jadeja', 'Shubman Gill', 'KL Rahul', 'Suryakumar Yadav',
        'Rishabh Pant', 'Jasprit Bumrah', 'Sanju Samson', 'Shikhar Dhawan'
    ],
    'total_runs': [7500, 5200, 6600, 2500, 2900, 3200, 4600, 3600, 3000, 400, 4000, 6700],
    'instagram_followers_millions': [268.0, 48.5, 32.5, 27.2, 8.1, 13.5, 17.0, 15.2, 10.4, 13.8, 4.5, 17.8]
}

df = pd.DataFrame(data)

# Calculate Spearman correlation
corr, p_value = spearmanr(df['total_runs'], df['instagram_followers_millions'])

print("Dataset:")
print(df.to_string(index=False))
print(f"\nSpearman Correlation: {corr:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Create a sample DataFrame simulating Flipkart products with price, user_rating, and num_reviews
n_products = 50
price = np.random.exponential(scale=2000, size=n_products) + 299
user_rating = np.random.uniform(1.0, 5.0, size=n_products)
num_reviews = np.random.poisson(lam=150, size=n_products) + 5

df = pd.DataFrame({
    'price': price,
    'user_rating': user_rating,
    'num_reviews': num_reviews
})

# Calculate the correlation matrix using pandas corr() method
corr_matrix = df[['price', 'user_rating', 'num_reviews']].corr()

print(corr_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'corr_matrix' is your DataFrame from the previous step
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix, 
    annot=True,          # Adds the correlation values inside the cells
    cmap='coolwarm',     # Color map ranging from blue (negative) to red (positive)
    fmt='.4f',           # Formats the annotations to 4 decimal places
    vmin=-1, vmax=1,     # Anchors the color bar scale between -1 and 1
    linewidths=0.5       # Adds grid lines separating the cells
)

plt.title('Flipkart Product Features Correlation Matrix', fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with the expected columns
df = pd.DataFrame({
    'subscribers': [1000, 5000, 20000, 150000, 2000000],
    'views': [5000, 25000, 80000, 900000, 15000000]
})

# Now the log transformation will work without a KeyError
df['log_subscribers'] = np.log1p(df['subscribers'])
df['log_views'] = np.log1p(df['views'])

linear_corr = df['log_subscribers'].corr(df['log_views'])
print(f"Linearized Pearson Correlation: {linear_corr:.4f}")
```
