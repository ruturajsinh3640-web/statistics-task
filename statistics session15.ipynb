```python
import pandas as pd
import numpy as np

# Read the CSV file
df = pd.read_csv("purchase_data.csv")

# Overall mean
overall_mean = df["purchase_value"].mean()

# Group by age_group
groups = [group["purchase_value"].values for name, group in df.groupby("age_group")]
group_names = [name for name, group in df.groupby("age_group")]

# Number of groups (k) and total observations (N)
k = len(groups)
N = len(df)

# Between-group variance (Mean Square Between, MSB = SSB / (k - 1))
ss_between = sum(len(g) * (g.mean() - overall_mean)**2 for g in groups)
df_between = k - 1
ms_between = ss_between / df_between

# Within-group variance (Mean Square Within, MSW = SSW / (N - k))
ss_within = sum(sum((g - g.mean())**2) for g in groups)
df_within = N - k
ms_within = ss_within / df_within

print(f"Between-group variance (MSB): {ms_between}")
print(f"Within-group variance (MSW): {ms_within}")
```
