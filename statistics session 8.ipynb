```python
import random

# Set a seed for reproducibility (optional)
random.seed(42)

# Parameters
p = 0.85
n_trials = 20

# Run the simulation (1 = success, 0 = failure)
results = [1 if random.random() < p else 0 for _ in range(n_trials)]

print("Simulation Results:", results)
print("Total Successes:", sum(results))
print(f"Observed Success Rate: {sum(results) / n_trials * 100}%")
```

```python
from scipy.stats import binom

def calculate_upi_probability():
    """
    Calculates the probability of exactly 3 successful UPI payments 
    out of 5 attempts, with a success rate of 0.9 per payment.
    """
    n = 5       # Total number of attempts
    p = 0.9     # Success rate per payment
    k = 3       # Exact number of successful payments
    
    # Calculate probability mass function
    probability = binom.pmf(k, n, p)
    
    return probability

# Example usage:
prob = calculate_upi_probability()
print(f"The probability of exactly 3 successes is: {prob:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
lam = 10       # Average reviews per hour
num_hours = 12 # Number of hours to simulate

# Simulate data using Poisson distribution
np.random.seed(42)  # Optional: for reproducibility
reviews = np.random.poisson(lam, num_hours)
hours = np.arange(1, num_hours + 1)

# Plotting the results
plt.figure(figsize=(10, 5))
plt.bar(hours, reviews, color='skyblue', edgecolor='navy', alpha=0.8)
plt.axhline(y=lam, color='red', linestyle='--', label=f'Expected Average ($\lambda$ = {lam})')

plt.title('Simulated Flipkart Product Reviews Over 12 Hours', fontsize=14, fontweight='bold')
plt.xlabel('Hour Index', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.xticks(hours)
plt.grid(axis='y', linestyle=':', alpha=0.7)
plt.legend()

plt.tight_layout()
plt.show()
```
